```python
# Import libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Load the dataset
df=pd.read_csv('Messy_data.csv')
```

```python
# Missing values
print(df.isnull().sum())
```

```python
# Remove Duplicate records
print(df.duplicated().sum())
```

```python
df=df.drop_duplicates()
```

```python
print(df.duplicated().sum())
```

```python
df.shape
```

```python
# Clean extra spaces and standarize the text
text_columns=["CustomerName","Product","City","OrderStatus"]
df["CustomerName"]=df["CustomerName"].str.strip()
df["CustomerName"]=df["CustomerName"].str.title()
df["Product"]=df["Product"].str.strip()
df["Product"]=df["Product"].str.title()
df["City"]=df["City"].str.strip()
df["City"]=df["City"].str.title()
df["OrderStatus"]=df["OrderStatus"].str.strip()
df["OrderStatus"]=df["OrderStatus"].str.title()
```

```python
for col in text_columns:
  df[col]=df[col].astype("string").str.strip().str.title()
```

```python
# Outliers
df["UnitPrice"]=pd.to_numeric(df["UnitPrice"],errors="coerce")
Q1=df["UnitPrice"].quantile(0.25)
Q3=df["UnitPrice"].quantile(0.75)
IQR=Q3-Q1
lower_bound=Q1-1.5*IQR
upper_bound=Q3+1.5*IQR
```

```python
# Box plot before handling outliers
plt.figure(figsize=(8, 4))
plt.boxplot(df["UnitPrice"].dropna())
plt.title("UnitPrice Before Outlier Handling")
plt.ylabel("Unit Price")
plt.show()
```

```python
df['UnitPrice']=df["UnitPrice"].clip(lower_bound,upper_bound)
```

```python
# Box plot after handling outliers
plt.figure(figsize=(8, 4))
plt.boxplot(df["UnitPrice"].dropna())
plt.title("UnitPrice After Outlier Handling")
plt.ylabel("Unit Price")
plt.show()
```

```python
# Create total amount
df["Quantity"]=pd.to_numeric(df["Quantity"],errors="coerce")
df["Total_Amount"]=df["Quantity"]*df["UnitPrice"]
df["OrderDate"]=pd.to_datetime(df["OrderDate"],errors="coerce")
```

```python
print(df['Total_Amount'].head())
```

```python
# Missing values
df["CustomerName"].fillna("Unknown",inplace=True)
df["CustomerAge"]=pd.to_numeric(df["CustomerAge"],errors="coerce")
df["CustomerAge"].fillna(df["CustomerAge"].median(),inplace=True)
df["City"].fillna("Unknown",inplace=True)
df["OrderStatus"].fillna("Unknown",inplace=True)
df["Quantity"]=df['Quantity'].fillna(df['Quantity'].median())
df["OrderDate"]=df['OrderDate'].fillna(df['OrderDate'].median())
df["UnitPrice"]=df['UnitPrice'].fillna(df['UnitPrice'].median())
df["Total_Amount"]=df['Total_Amount'].fillna(df['Total_Amount'].median())

```

```python
# Check whether Quantity,UnitPrice and CustomerAge contain any negative values
print(df[df["Quantity"] <= 0])

print(df[df["UnitPrice"] < 0])

print(df[df["CustomerAge"] < 0])

print(df[df["Total_Amount"] <0])
```
